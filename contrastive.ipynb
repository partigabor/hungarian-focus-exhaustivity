```python
import pandas as pd

df = pd.read_excel("contrastive_results.xlsx")
print(df.columns.tolist())
```

```python
# Rename 'Anyanyelve:' to 'native_language' and '18 éves koráig tanulmányait Magyarországon végezte?' to 'schooling', 'Neme:' to 'gender', and 'Kora:' to 'age'. 
df.rename(
    columns={
        'Anyanyelve:': 'native_language',
        '18 éves koráig tanulmányait Magyarországon végezte?': 'schooling',
        'Neme:': 'gender',
        'Kora:': 'age'
    }, inplace=True
)

print(df.columns.tolist())
```

```python

# Show the results for these 4 columns one by one is a simple and meaningful way.
print("Native Language:")
print(df['native_language'].value_counts())
print()

print("Schooling:")
print(df['schooling'].value_counts())
print()

print("Gender:")
print(df['gender'].value_counts())
print()

print("Age:")
print(df['age'].value_counts())
print()
```

```python
# Calculate mean, and sd for age
print(f"Mean age: {df['age'].mean()}")
print(f"Standard deviation of age: {df['age'].std()}")
```

```python
df

# Transpose this table
df = df.T

# Drop Timestamp, native_language, schooling, gender, and age columns
df.drop(['Timestamp', 'native_language', 'schooling', 'gender', 'age'], axis=0, inplace=True)
df
```

```python
# Turn the index column into a column called 'sentence', and reindex the df simply
df.reset_index(inplace=True)
df.rename(columns={'index': 'sentence'}, inplace=True)
df
```

```python
# Read in the contrastive_sentences.xlsx for reference
reference_df = pd.read_excel('contrastive_sentences.xlsx')

# Preserve original `df` row order and add a `type` column where sentences match.
# This maps each sentence to its `type` value from the reference, leaving unmatched rows as NaN.
if 'sentence' in reference_df.columns and 'type' in reference_df.columns:
    type_map = reference_df.set_index('sentence')['type'].to_dict()
    df['type'] = df['sentence'].map(type_map)
    # ensure `type` is the last column
    cols = [c for c in df.columns if c != 'type'] + ['type']
    df = df[cols]
else:
    print("reference_df must contain 'sentence' and 'type' columns. Found:", reference_df.columns.tolist())

# show resulting frame
df
```

```python
# Convert rating columns to numeric and calculate item-level statistics
rating_cols = [col for col in df.columns if col != "type"]

ratings = df[rating_cols].apply(pd.to_numeric, errors="coerce")

item_stats = pd.DataFrame({
    "mean": ratings.mean(axis=1),
    "sd": ratings.std(axis=1, ddof=1),
    "n": ratings.count(axis=1)
})

item_stats
```

```python
import plotly.express as px

plot_df = item_stats.reset_index().rename(columns={"index": "sentence"})
# drop the non-sentence metadata row and align sentence/type by row position
plot_df = plot_df.iloc[1:].reset_index(drop=True).copy()
plot_df["sentence"] = df["sentence"].iloc[1:].reset_index(drop=True).to_numpy()
plot_df["type"] = df["type"].iloc[1:].reset_index(drop=True).to_numpy()
plot_df["item"] = range(1, len(plot_df) + 1)

fig = px.scatter(
    plot_df,
    x="item",
    y="mean",
    color="type",
    error_y="sd",
    hover_data=["sentence", "mean", "sd", "n"],
    labels={
        "item": "Item",
        "mean": "Mean rating",
        "type": "Sentence type",
        "sd": "Standard deviation",
    },
    title="Mean Ratings by Sentence Type",
)

fig.update_traces(marker={"size": 9})
fig.update_layout(
    xaxis={"dtick": 1},
    yaxis={"range": [1, 7]},
)
fig.show()

# Export this as png
fig.write_image("mean_ratings_by_sentence_type.png")
```

```python
# Calculate the mean and standard deviation of ratings for each sentence type
type_stats = plot_df.groupby("type").agg(
    mean=("mean", "mean"),
    sd=("sd", "std"),
)

# Print the targeted statistics for each sentence type
print("Target sentence type statistics:")
for sentence_type, stats in type_stats.iterrows():
    print(f"Type: {sentence_type}, Mean: {stats['mean']:.2f}, SD: {stats['sd']:.2f}")
```
