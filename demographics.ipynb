# Prolific demographics summary
Collect all exported demographic CSV files, keep only accepted participants, and compute age and gender summaries.

```python
from pathlib import Path
import pandas as pd
import plotly.express as px

data_dir = Path("demographic_data")
csv_files = sorted(data_dir.glob("*.csv"))
if not csv_files:
    raise FileNotFoundError(f"No CSV files found in {data_dir.resolve()}")

raw = pd.concat((pd.read_csv(path) for path in csv_files), ignore_index=True)
raw.head()
```

```python
raw
```

```python
accepted = (
    raw[raw["Status"].str.upper() == "APPROVED"]
    .assign(Age=lambda df: pd.to_numeric(df["Age"], errors="coerce"))
    .sort_values("Completed at")
    .drop_duplicates(subset="Participant id", keep="last")
    .reset_index(drop=True)
 )
accepted[["Participant id", "Status", "Age", "Sex"]].head()
```

```python
accepted
```

```python
list_of_participants_to_remove = ['642b35c70771761602e9c3ae', '641379405684937e6fad9f1b', '5f046bf88d2c186cc10c6ad0']
accepted = accepted[~accepted["Participant id"].isin(list_of_participants_to_remove)].reset_index(drop=True)
```

```python
accepted
```

```python
age_summary = accepted["Age"].describe().to_frame().T
gender_summary = (
    accepted["Sex"]
    .str.title()
    .value_counts(dropna=False)
    .rename_axis("Sex")
    .to_frame("count")
 )
gender_summary["percent"] = (gender_summary["count"] / gender_summary["count"].sum() * 100).round(1)

total_participants = len(accepted)
age_summary, gender_summary, total_participants
```

```python
mean_age = age_summary.loc["Age", "mean"]
std_age = age_summary.loc["Age", "std"]
min_age = age_summary.loc["Age", "min"]
max_age = age_summary.loc["Age", "max"]

gender_parts = ", ".join(
    f"{sex.lower()} {row['percent']:.1f}% (n={row['count']})"
    for sex, row in gender_summary.iterrows()
)

summary_note = (
    f"The analytic sample comprised {total_participants} approved Prolific participants "
    f"(M_age = {mean_age:.1f} years, SD = {std_age:.1f}, range = {min_age:.0f}–{max_age:.0f}). "
    f"Gender composition was {gender_parts}."
)
print(summary_note)
```

## Plotly donut charts
Interactive visuals to quickly scan gender distribution and age-band composition among accepted participants.

```python
gender_fig = px.pie(
    gender_summary.reset_index(),
    names="Sex",
    values="count",
    hole=0.4,
    title=f"Gender distribution (accepted participants, n={total_participants})",
    color_discrete_sequence=px.colors.qualitative.Pastel,
)
# balance label visibility for print/export
gender_fig.update_traces(
    texttemplate="<b>%{label}</b><br>n=%{value}<br>%{percent}",
    textfont=dict(size=24, family="Times New Roman"),
    hovertemplate="%{label}: %{value} participants",
)
# square canvas so the two donuts sit side-by-side cleanly
gender_fig.update_layout(
    width=700,
    height=700,
    margin=dict(l=20, r=20, t=60, b=20),
    font=dict(size=24, family="Times New Roman"),
    showlegend=False,
)

# Save fig to plots (square export)
gender_fig.write_image("plots/demographics_gender_pie.png", width=1400, height=1400, scale=3)

# show fig
gender_fig
```

```python
age_bins = pd.cut(
    accepted["Age"],
    bins=[18, 25, 30, 35, 40, 50, 60, 69],
    labels=["18-24", "25-29", "30-34", "35-39", "40-49", "50-59", "60+"],
    right=False,
)

# enforce logical ascending order for plotting
age_order = ["18-24", "25-29", "30-34", "35-39", "40-49", "50-59", "60+"]
age_counts = (
    age_bins.value_counts(sort=False)
    .reindex(age_order, fill_value=0)
    .reset_index()
)
age_counts.columns = ["Age band", "count"]

age_fig = px.pie(
    age_counts,
    names="Age band",
    values="count",
    hole=0.5,
    title=f"Age bands (accepted participants, n={total_participants})",
    color_discrete_sequence=px.colors.qualitative.Set3,
    category_orders={"Age band": age_order},
)
age_fig.update_traces(
    texttemplate="<b>%{label}</b><br>n=%{value}<br>%{percent}",
    textfont=dict(size=24, family="Times New Roman"),
    hovertemplate="%{label}: %{value} participants",
)

# square canvas to align with gender chart
age_fig.update_layout(
    width=700,
    height=700,
    margin=dict(l=20, r=20, t=60, b=20),
    font=dict(size=24, family="Times New Roman"),
)

# Save fig to plots (square export)
age_fig.write_image("plots/demographics_age_pie.png", width=1400, height=1400, scale=3)

# show fig
age_fig
```

```python
from plotly.subplots import make_subplots
import plotly.graph_objects as go

gender_colors = px.colors.qualitative.Pastel
age_colors = px.colors.qualitative.Set3

combined_fig = make_subplots(
    rows=1,
    cols=2,
    specs=[[{"type": "domain"}, {"type": "domain"}]],
    subplot_titles=[
        f"Gender (n={total_participants})",
        f"Age bands (n={total_participants})",
    ],
)

combined_fig.add_trace(
    go.Pie(
        labels=gender_summary.index,
        values=gender_summary["count"],
        hole=0.4,
        marker=dict(colors=gender_colors),
        texttemplate="<b>%{label}</b><br>n=%{value}<br>%{percent}",
        textfont=dict(size=24, family="Times New Roman"),
        hovertemplate="%{label}: %{value} participants",
    ),
    row=1,
    col=1,
)

combined_fig.add_trace(
    go.Pie(
        labels=age_counts["Age band"],
        values=age_counts["count"],
        hole=0.4,
        marker=dict(colors=age_colors),
        texttemplate="<b>%{label}</b><br>n=%{value}<br>%{percent}",
        textfont=dict(size=24, family="Times New Roman"),
        hovertemplate="%{label}: %{value} participants",
    ),
    row=1,
    col=2,
)

combined_fig.update_layout(
    width=1200,
    height=500,
    margin=dict(l=0, r=0, t=40, b=0),
    font=dict(size=24, family="Times New Roman"),
    showlegend=False,
)

for ann in combined_fig.layout.annotations:
    ann.font.size = 24
    ann.font.family = "Times New Roman"

combined_fig.show()

combined_fig.write_image("plots/demographics_combined_donuts.png", scale=2)

```

```python
# Approximate mean age using band midpoints
age_band_midpoints = {
    "18-24": 21,
    "25-29": 27,
    "30-34": 32,
    "35-39": 37,
    "40-49": 44.5,
    "50-59": 54.5,
    # upper bin spans 60–69 (changed upper cut to 69); use its midpoint
    "60+": 64.5,
}

age_band_mean = (
    age_counts.assign(midpoint=lambda df: df["Age band"].map(age_band_midpoints))
    .pipe(lambda df: (df["count"] * df["midpoint"]).sum() / df["count"].sum())
)

print(f"Mean age (band midpoint weighted): {age_band_mean:.1f} years")
```
